```python
#Creating Arrays Using NumPy Functions
#np.zeros()

arr = np.zeros((3,3))
print(arr)
```

```python
#np.ones()

arr = np.ones((4,5))
print(arr)
```

```python
#np.arange()
arr = np.arange(2,20,2)
print(arr)
```

```python
#Array Operations
#ADDITION

A = np.array([45,50,70])
B = np.array([20,10,5])
print(A+B)
```

```python
#SUBSTRACTION

A = np.array([45,50,70])
B = np.array([20,10,5])
print(A-B)
```

```python
#MULTIPLICATION

A = np.array([45,50,70])
B = np.array([20,10,5])
print(A*B)
```

```python
#DIVISION

A = np.array([45,50,70])
B = np.array([20,10,5])
print(A/B)

```

```python
#SQUARE

A = np.array([3,4,5])
print(A**2)
```

```python
#Common NumPy Functions

#np.sum()
sales = np.array([12000,15000,18000])
print(np.sum(sales))
```

```python
#np.mean()

sales = np.array([12000,15000,18000])
print(np.mean(sales))
```

```python
#np.max()

salary = np.array([25000,32000,28000])
print(np.max(salary))
```

```python
#np.min()

salary = np.array([25000,32000,28000])
print(np.min(salary))
```

```python
#np.std()

marks = np.array([70,80,90,100])
print(np.std(marks))
```

```python
#np.var()

marks = np.array([70,80,90,100])
print(np.var(marks))
```

```python
# np.sqrt()

numbers = np.array([4,9,16,25])
print(np.sqrt(numbers))
```

```python
# np.sort()

numbers = np.array([25,10,30,15])
print(np.sort(numbers))
```

```python
#np.unique()

numbers = np.array([1,2,2,3,4,4,5])
print(np.unique(numbers))
```

```python
#np.reshape()

arr = np.array([1,2,3,4,5,6])
print(arr.reshape(2,3))
```

```python
#np.concatenate()

a = np.array([1,2,3])
b = np.array([4,5,6])
print(np.concatenate((a,b)))
```

```python
#np.argmax()

marks = np.array([78,85,95,88])
print(np.argmax(marks))
```

```python
#np.argmin()

marks = np.array([78,85,95,88])
print(np.argmin(marks))
```
