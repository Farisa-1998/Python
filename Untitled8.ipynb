```python
#CONTROL STATEMENTS

#While Loop - repeats a block of code as long as the condition is True

i = 3

while i <= 20:
    print(i)
    i += 3
```

```python
#BREAK STATEMENTS

#while condition:if condition: break


i = 1

while i <= 20:

    if i == 14:
        break

    print(i)
    i += 1
```
