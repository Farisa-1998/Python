```python
#Python DA Assignment 1: Data Analysis using NumPy and Pandas

#Numpy Array Operations:

#1. Create a 1D NumPy Array:

import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python
#2. Inspection and Properties:

print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
#3. Array Operations:

fahrenheit = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", fahrenheit)
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
#4. Array Slicing and Indexing:

print("First three days:", temperatures_w1[:3])

print("Weekend temperatures:", temperatures_w1[-2:])

print("Middle three days:", temperatures_w1[2:5])
```

```python
#5. Create a 2D Array:

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

```python
#6. Inspect and Slice the 2D Array:

print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)

print("Week 1 Temperatures:", temperatures[0])
print("Week 2 Temperatures:", temperatures[1])

print("Week 1 Weekend Temperatures:", temperatures[0, -2:])
print("Week 2 Weekend Temperatures:", temperatures[1, -2:])

```

```python
#Pandas Series:

#1. Creating Pandas Series:

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
#2. Indexing and Slicing:

print("1st Rank Mark:", marks.iloc[0])

print("\nTop 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

print("\n3rd Rank Mark:", marks.iloc[2])

print("\nMarks Greater Than 90:")
print(marks[marks > 90])
```

```python
#3. Manipulating Series:

# Modify Rank1 mark
marks.loc['Rank1'] = 100

print("After modifying Rank1:")
print(marks)

# Remove Rank5
marks = marks.drop('Rank5')

print("\nAfter removing Rank5:")
print(marks)

# Calculate CGPA
cgpa = marks / 10

print("\nCGPA:")
print(cgpa)
```

```python
#Pandas DataFrame:

#Creating Pandas DataFrame:

data = { 'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]}

transactions = pd.DataFrame(data)

print(transactions)
```

```python
#2. Data Exploration:

# Display DataFrame
print("Full DataFrame:")
print(transactions)

# First 5 rows
print("\nHead:")
print(transactions.head())

# Last 5 rows
print("\nTail:")
print(transactions.tail())

# Shape
print("\nShape:")
print(transactions.shape)

# Column names
print("\nColumn Names:")
print(transactions.columns)

# Data types
print("\nData Types:")
print(transactions.dtypes)

# Display ProductCategory and Amount
print("\nProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])

# Retrieve last 3 columns
print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])

# Filter North region and Amount greater than 200
print("\nNorth Region with Amount Greater Than 200:")
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])

# Value counts
print("\nProduct Category Value Counts:")
print(transactions['ProductCategory'].value_counts())

# Unique regions
print("\nUnique Regions:")
print(transactions['Region'].unique())

# Mean Amount by Region
print("\nMean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
#Manipulating the DataFrame:

# Modify Amount for TransactionID 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("After modifying Amount:")
print(transactions)

# Add Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

print("\nAfter adding Discount:")
print(transactions)

# Remove TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

print("\nAfter removing TransactionID 109:")
print(transactions)

# Delete Discount column
transactions = transactions.drop('Discount', axis=1)

print("\nFinal DataFrame:")
print(transactions)
```
