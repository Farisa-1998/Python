```python
#DICTIONARY -  unordered , mutable collection of key-value pairs.

student = {
    "Name": "Farisa",
    "Age": 28,
    "Course": "Data Analytics",
    "Marks": 98
}
print(student)
```

```python
#METHODS IN DICTIONARY
#GET - Returns the value of a key.

student = {"Name":"Farisa","Age":28}

print(student.get("Name"))
```

```python
#KEYS - Returns all keys.

student = {"Name":"Farisa","Age":28}

print(student.keys())
```

```python
#VALUES - Returns all values

student = {"Name":"Farisa","Age":28}
print(student.values())
```

```python
#ITEMS - Returns all key-value pairs.

student = {"Name":"Farisa","Age":28}
print(student.items())
```

```python
#UPDATE - Updates the dictionary

student = {"Name":"Farisa"}
student.update({"Age":28})
print(student)
```

```python
#POP

student = {"Name":"Farisa","Age":28}
student.pop("Age")
print(student)
```

```python
#POP ITEM -

student = {"Name":"Farisa","Age":28}
student.popitem()
print(student)
```

```python
#CLEAR -

student = {"Name":"Farisa","Age":28}
student.clear()
print(student)
```

```python
#COPY -

student = {"Name":"Farisa","Age":28}
new_student = student.copy()
print(new_student)

```

```python
#SET -unordered collection of unique elements

fruits = {"Banana", "Kiwi", "Litchi"}
print(fruits)
```

```python
#METHODS OF SET
#ADD - Adds one element.

fruits = {"Banana", "Kiwi","Litci"}
fruits.add("Mango")
print(fruits)

```

```python
#UPDATE - Adds multiple items.

fruits = {"Apple"}
fruits.update(["Orange","Mango"])
print(fruits)
```

```python
#REMOVE - Removes an item.

fruits = {"Banana", "Kiwi","Litci"}
fruits.remove("Kiwi")
print(fruits)
```

```python
# discard- Removes an item if present.

fruits = {"Banana", "Kiwi","Litci"}
fruits.remove("Mango")
print(fruits)
```

```python
#pop()-Removes a random item.


fruits = {"Banana", "Kiwi","Litci"}
fruits.pop()
print(fruits)

```

```python
#clear()- Removes all elements.

fruits = {"Banana", "Kiwi","Litci"}
fruits.clear()
print(fruits)
```

```python
#copy() - Returns a copy of the set.

fruits = {"Banana", "Kiwi","Litci"}
fruits.copy()
print(fruits)

```

```python
# union() - Combines two sets.

A = {100,300,500,700,900}
B = {200,400,600,800,1000}
print(A.union(B))
```

```python
# intersection() - Returns common elements.

A = {100,300,500,700,900}
B = {100,400,300,800,700}
print(A.intersection(B))
```

```python
# difference() -Returns elements present only in the first set.

A = {100,300,500,700,900}
B = {100,400,300,800,700}
print(A.difference(B))
```

```python
A = {100,300,500,700,900}
B = {100,400,300,800,700}
print(B.difference(A))
```

```python
#symmetric_difference()- Returns elements that are in either set but not in both.

A = {100,300,500,700,900}
B = {100,400,300,800,700}
print(B.symmetric_difference(A))





```
