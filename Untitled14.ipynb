```python
#Python DA Assignment 2 - Data Visualization

#1) Loading the Taxis Dataset:

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Display the first five rows
df.head()
```

```python
#Handling Missing Values:

# Check missing values in each column
df.isnull().sum()
```

```python
# Total missing values in the dataset
df.isnull().sum().sum()
```

```python
# Identify columns with missing values
missing_columns = df.columns[df.isnull().any()]

print("Columns with missing values:")
print(missing_columns)
```

```python
# Fill missing numerical values with median

for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Fill missing categorical values with mode
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Remove rows with remaining missing values
df.dropna(inplace=True)

# Check missing values again
df.isnull().sum()
```

```python
#3) Visualizations using Matplotlib/Pandas Plot:

#3.1 Line Chart – Fare Over Time

# Convert pickup column to datetime
df['pickup'] = pd.to_datetime(df['pickup'])

# Sort data by pickup timestamp
df = df.sort_values(by='pickup')
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df['pickup'], df['fare'], color='blue')

plt.title('Fare Over Time')
plt.xlabel('Pickup Timestamp')
plt.ylabel('Fare Amount')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#3.2 Bar Chart – Total Fare by Pickup Borough

# Group data by pickup borough and calculate total fare
borough_fare = df.groupby('pickup_borough')['fare'].sum()

# Create bar chart
borough_fare.plot(kind='bar', figsize=(8, 5), color='orange')

plt.title('Total Fare by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Total Fare')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#3.3 Pie Chart – Trips by Payment Method

# Count trips for each payment method
payment_counts = df['payment'].value_counts()

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribution of Trips by Payment Method')
plt.show()
```

```python
#3.4 Histogram – Distribution of Distance

plt.figure(figsize=(9, 5))

plt.hist(df['distance'], bins=20, edgecolor='black')

plt.title('Distribution of Taxi Trip Distance')
plt.xlabel('Distance')
plt.ylabel('Number of Trips')

plt.tight_layout()
plt.show()
```

```python
#3.5 Box Plot – Tip Amount by Pickup Borough

plt.figure(figsize=(10, 6))

df.boxplot(column='tip', by='pickup_borough')

plt.title('Tip Amount by Pickup Borough')
plt.suptitle('')
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')

plt.tight_layout()
plt.show()
```

```python
#4.Visualizations using Seaborn:

#4.1 Count Plot – Number of Trips by Pickup Borough

plt.figure(figsize=(9, 5))

sns.countplot(data=df, x='pickup_borough')

plt.title('Number of Trips by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Count of Trips')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#4.2 Scatter Plot – Distance vs Fare

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='distance',
    y='fare',
    hue='pickup_borough'
)

plt.title('Relationship Between Distance and Fare')
plt.xlabel('Distance')
plt.ylabel('Fare')

plt.tight_layout()
plt.show()
```

```python
#4.3 Heatmap – Correlation Between Numerical Variables

# Select numerical columns
numeric_data = df[['distance', 'fare', 'tip', 'tolls', 'total']]

# Calculate correlation matrix
correlation_matrix = numeric_data.corr()

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
#4.4 Pair Plot – Pairwise Relationships

# Select required columns
pair_data = df[['distance', 'fare', 'tip', 'total', 'pickup_zone']]

# Create pair plot
sns.pairplot(
    pair_data,
    hue='pickup_zone'
)

plt.show()
```

```python
#4.5 Violin Plot – Fare Distribution by Payment Method

plt.figure(figsize=(9, 6))

sns.violinplot(
    data=df,
    x='payment',
    y='fare'
)

plt.title('Fare Distribution by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Fare Amount')

plt.tight_layout()
plt.show()
```
