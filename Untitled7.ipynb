```python
#Python Assignment 2: Data Structures - List, Dictionary, Set & Conditional Statements:

#List (Creation, Modification and Access):

# 1(a). Create a list of five integers
age_list = [24, 25, 32, 27, 28]
print("Age List:", age_list)

# 1(b). Create a list of five strings
name_list = ["Farisa", "Anam", "Faris", "Faras", "Shanif"]
print("Name List:", name_list)
```

```python
# a. Append Yazhini to name_list
name_list.append("Yazhini")
print("After append:", name_list)

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("After insert:", age_list)

# c. Remove Yazhini from name_list
name_list.remove("Yazhini")
print("After remove:", name_list)

# d. Pop the last element from age_list
age_list.pop()
print("After pop:", age_list)

# e. Extend age_list with additional ages
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("Descending order:", age_list)

# g. Find maximum, minimum and sum
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
# 3. Accessing List Elements

# a. Print the first element of name_list
print("First name:", name_list[0])

# b. Print the last element of name_list
print("Last name:", name_list[-1])

# c. Print elements from index 2 to index 4
print("Names from index 2 to 4:", name_list[2:5])

# d. Print the elements in reverse order
print("Reverse order:", name_list[::-1])
```

```python
#Dictionary (Creation, Modification and Access):

# 1. Create a dictionary of five students and their marks

student_marks = {"Hasna": 75,
                 "Anam": 88,
                 "Shanif": 67,
                 "Farisa": 91,
                 "Nafih": 78}
print("Student Marks:", student_marks)

# 2. Access and print a specific student's mark
print("Anam's mark:", student_marks["Anam"])

# 3. Add a new student named Janani
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# 4. Update an existing student's mark to 82
student_marks["Faris"] = 82
print("After updating Faris:", student_marks)
```

```python
# 5. Display all keys, values, and key-value pairs

# Display all student names
print("Student names:", student_marks.keys())

# Display all marks
print("Student marks:", student_marks.values())

# Display all key-value pairs
print("Student name and marks:", student_marks.items())
```

```python
#Sets (Operations):

# 1. Create a set with vowels
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print("My Set:", my_set)
print("Number of elements:", len(my_set))
```

```python
# Attempt to change an element using an index
my_set[4] = 's'

#Explanation: Sets are unordered collections and do not support indexing or item assignment. Therefore, we cannot change an element using my_set[4].
```

```python
# Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# Find the union
print("Union:", set1.union(set2))

# Find the intersection
print("Intersection:", set1.intersection(set2))
```
